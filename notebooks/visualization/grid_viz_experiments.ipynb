```python
import uxarray as ux

import geoviews as gv
import geoviews.feature as gf

import cartopy.crs as ccrs

import numpy as np
import xarray as xr
```

```python
# 
# grid_fil = "/Users/brianpm/Downloads/mpasa240_SCRIP_desc_211115.nc"

grid_fil = "/Users/brianpm/Downloads/mpasa60_SCRIP_desc_211009.nc"
```

```python

# primal_mesh = ux.open_grid(grid_fil, use_dual=False)
# dual_mesh = ux.open_grid(grid_fil, use_dual=True)

ds = xr.open_dataset(grid_fil)
ds['grid_center_lat'] = np.rad2deg(ds['grid_center_lat'])
ds['grid_center_lon'] = np.rad2deg(ds['grid_center_lon'])
ds['grid_corner_lon'] = np.rad2deg(ds['grid_corner_lon'])
ds['grid_corner_lat'] = np.rad2deg(ds['grid_corner_lat'])

# save as new file
fixfile = "/Users/brianpm/Downloads/mpasa60_SCRIP_desc_INDEGREES.nc"
ds.to_netcdf(fixfile)
```

```python
grid_fil = fixfile
uxds = ux.open_dataset(grid_fil, grid_fil)
grid = uxds.uxgrid
```

```python
r = grid.plot.mesh(backend='matplotlib', exclude_antimeridian=True)
# gv.save(r, "/Users/brianpm/Desktop/example.png")
r
```

```python
uxds['grid_area'].plot.polygons(backend='matplotlib', title="Plot", height=350, width=700).opts(facecolor='none')
```

```python
gv.extension('matplotlib')

gv.output(size=1250)

polys = gv.Polygons(uxds['grid_area'].to_geodataframe())

r = (gf.coastline * gf.land * gf.ocean * polys.opts(alpha=0.25, cmap='YlOrBr_r', color='red', edgecolor='red', facecolor='none')).opts(projection=ccrs.Orthographic(central_longitude=-45, central_latitude=20), global_extent=True, backend='matplotlib')
gv.save(r, "/Users/brianpm/Desktop/example_mpas60.png")
r
```
