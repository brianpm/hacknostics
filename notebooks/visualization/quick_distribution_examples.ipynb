```python
import numpy as np 
import matplotlib.pyplot as plt 
from scipy.stats import gamma

x = np.linspace(0, 10, 1000) # first gamma function with shape=2 and mean=3 
y1 = gamma.pdf(x, 2, 3) # second gamma function with shape=4 and mean=6 
y2 = gamma.pdf(x, 4, 6) 
plt.plot(x, y1, label='Gamma(2,3)') 
plt.plot(x, y2, label='Gamma(4,6)') 
plt.xlabel('x') 
plt.ylabel('gamma(x)') 
plt.legend() 
plt.show()
```

```python
import numpy as np 
import matplotlib.pyplot as plt
# Generate random cloud droplet sizes 
num_cloud_droplets = 1000 
cloud_sizes = np.random.lognormal(2.5, 0.7, size=num_cloud_droplets)
# Generate random rain drop sizes 
num_rain_droplets = 500 
rain_sizes = np.random.lognormal(3.0, 0.8, size=num_rain_droplets)
# Plot the droplet sizes 
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1) 
plt.hist(cloud_sizes, bins=20, color='gray', alpha=0.5) 
plt.xlabel('Cloud Droplet Size') 
plt.ylabel('Frequency')
plt.subplot(1, 2, 2) 
plt.hist(rain_sizes, bins=20, color='blue', alpha=0.5) 
plt.xlabel('Rain Drop Size') 
plt.ylabel('Frequency')
plt.suptitle('Population of Cloud Droplets and Rain Drops')
plt.show()

# This code generates random sizes for a population of cloud droplets and rain drops using lognormal distributions, plots the size distributions in separate histograms with different colors, and shows the results side by side in one figure.
```

```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy.stats import gamma

# Generate random cloud droplet sizes using gamma distribution
num_cloud_droplets = 1000 
cloud_sizes = gamma.rvs(2, scale=3, size=num_cloud_droplets)

# Generate random rain drop sizes using gamma distribution
num_rain_droplets = 100
rain_sizes = gamma.rvs(4, scale=6, size=num_rain_droplets)

# draw the continuous distribution
x = np.linspace(0, 80, 1000) # first gamma function with shape=2 and mean=3 
y1 = gamma.pdf(x, 2, scale=3) # second gamma function with shape=4 and mean=6 
y2 = gamma.pdf(x, 4, scale=6) 

# frozen pdf
# y1rv = gamma(2)
# y2rv = gamma(4)
# y1 = y1rv.pdf(x)
# y2 = y2rv.pdf(x)

# Plot the droplet sizes in histograms
fig, ax = plt.subplots(figsize=(12,6), ncols=2)
ax[0].hist(cloud_sizes, bins=20, color='gray', alpha=0.5, label='Cloud', density=True)
ax[0].set_xlabel('Size')
ax[0].set_ylabel('Frequency')

ax[0].hist(rain_sizes, bins=20, color='blue', alpha=0.5, label='Rain', density=True)
# ax[0].set_xlabel('Rain Drop Size')
# ax[0].set_ylabel('Frequency')
ax[0].legend()

ax[0].plot(x, y1, color='black')
ax[0].plot(x, y2, color='blue')

# Create a scatter plot to randomly place the populations of droplets and drops in a 2D space
ax[1].scatter(np.random.rand(num_cloud_droplets), np.random.rand(num_cloud_droplets), s=cloud_sizes, color='gray', alpha=0.5, label='Cloud Droplets')
ax[1].scatter(np.random.rand(num_rain_droplets), np.random.rand(num_rain_droplets), s=rain_sizes, color='blue', alpha=0.5, label='Rain Drops')
ax[1].set_xlabel('X')
ax[1].set_ylabel('Y')
ax[1].legend()

fig.suptitle('Population of Cloud Droplets and Rain Drops with Gamma Distributions')
fig.show()
```
